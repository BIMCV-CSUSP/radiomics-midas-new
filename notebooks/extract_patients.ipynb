```python
#join csv
import pandas as pd
import os

path1 = '/mnt/datalake/openmind/MedP-Midas/sgonzalez/radiomics-midas-new/data/csv/patients_project_p0032025.csv'
path2 = '/mnt/datalake/openmind/MedP-Midas/sgonzalez/radiomics-midas-new/data/csv/patients_project_p0072025.csv'
path3 = '/mnt/datalake/openmind/MedP-Midas/sgonzalez/radiomics-midas-new/data/csv/patients_project_p0182025.csv'
path4 = '/mnt/datalake/openmind/MedP-Midas/sgonzalez/radiomics-midas-new/data/csv/patients_project_p0202025.csv'

df1 = pd.read_csv(path1)
df2 = pd.read_csv(path2)
df3 = pd.read_csv(path3)
df4 = pd.read_csv(path4)

df = pd.concat([df1, df2, df3, df4], ignore_index=True)

#save df to csv
# output_path = '/mnt/datalake/openmind/MedP-Midas/sgonzalez/radiomics-midas-new/data/csv/patients_project_combined.csv'
# df.to_csv(output_path, index=False)
```

```python
#delete patients with 'patient_age' NaN
df_clean = df.dropna(subset=['patient_age'])
```

```python
df_weeks = df[df['patient_age'].astype(str).str.upper() == '004W']
print(df_weeks)
# #substitue  '004W' with '4Y'
# df_clean['patient_age'] = df_clean['patient_age'].replace('004W', '4Y')
```

```python
#modify format age 044Y and 004W to 44 and 4 respectively
# df_clean['patient_age'] = df_clean['patient_age'].str.replace('W', '').str.replace('Y', '').str.zfill(3)
#convert to int
df_clean['patient_age'] = df_clean['patient_age'].astype(str).str.replace('Y', '').str.replace('W', '').str.zfill(3).astype(int)
```

```python
#filter patients with age < 30 or > 75
df_clean_filtered = df_clean[(df_clean['patient_age'] < 30) | (df_clean['patient_age'] > 75)]
# df_clean_filtered = df_clean[df_clean['patient_age'] <30]
```

```python
# delete columns study_description,series_descriptions,t2w_series,t2w_count,dicom_files_found and dicom_files_analyzed
df_clean_filtered = df_clean_filtered.drop(columns=['study_description', 'series_descriptions', 't2w_series', 't2w_count', 'dicom_files_found', 'dicom_files_analyzed'])
```

```python
# add empty columns with names : L5-S,L4-L5,L3-L4,L2-L3,L1-L2
df_clean_filtered['L5-S'] = ''
df_clean_filtered['L4-L5'] = ''
df_clean_filtered['L3-L4'] = ''
df_clean_filtered['L2-L3'] = ''
df_clean_filtered['L1-L2'] = ''
```
