```python
import pandas as pd

import ast
import numpy as np
```

```python
# Choose the best model BINARY
df = pd.read_csv("/mnt/datalake/openmind/MedP-Midas/sgonzalez/radiomics-midas-new/binary/results_lumbar_discs_label.csv")
# val_f1_macro mean 
best_model = df.loc[df['val_auc'].idxmax()]

#read only the features of the model named Gradient Boosting
gb_data = df[df["Classifier"] == "Gradient Boosting"]

val_mean = gb_data["val_auc"].mean()
val_std = gb_data["val_auc"].std()
print(f"Gradient Boosting - val_auc mean: {val_mean}, std: {val_std}")


all_clases_elem = []
for f1_str in gb_data["per_class_f1"]:
    # Convertir string a lista
    f1_list = np.array(ast.literal_eval(f1_str))
    all_clases_elem.append(f1_list)

all_clases_elem = np.array(all_clases_elem)

all_clases_elem.mean(axis=0)
all_clases_elem.std(axis=0)
print(f"Gradient Boosting - per_class_f1 mean: {all_clases_elem.mean(axis=0)}, std: {all_clases_elem.std(axis=0)}")
```
